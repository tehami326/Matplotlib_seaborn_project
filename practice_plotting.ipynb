```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 5, 11)
y = x**3

plt.plot(x, y)
plt.title("Plotting")
plt.xlabel("X axis")
plt.ylabel("Y axis")



plt.show()
```

```python
plt.subplot(2,2,1)
plt.subplot(2,2,2)
plt.subplot(2,2,3)
plt.subplot(2,2,4)
plt.plot(x,y)
```

```python
fig = plt.figure(figsize=(6,2),dpi=100)
axis1 = fig.add_axes([0.1,0.1,0.6,0.8])
axis1.plot(x,y)
```

```python
plt.scatter(x,y)
```
